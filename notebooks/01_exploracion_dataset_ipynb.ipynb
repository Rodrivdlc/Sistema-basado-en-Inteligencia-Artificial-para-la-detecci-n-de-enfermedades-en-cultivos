```python
dataset_path = "/content/dataset/PlantVillage"

print(os.listdir(dataset_path)[:20])
```

```python
clases = os.listdir(dataset_path)

clases_tomate = [
    clase for clase in clases
    if clase.lower().startswith("tomato")
]

print("Clases de tomate encontradas:")
for clase in clases_tomate:
    print("-", clase)
```

```python
print("Cantidad de imágenes por clase:\n")

for clase in clases_tomate:
    ruta_clase = os.path.join(dataset_path, clase)

    cantidad = len([
        archivo for archivo in os.listdir(ruta_clase)
        if archivo.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    print(f"{clase}: {cantidad}")
```

```python
cantidades = []

for clase in clases_tomate:
    ruta_clase = os.path.join(dataset_path, clase)

    cantidad = len([
        archivo for archivo in os.listdir(ruta_clase)
        if archivo.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    cantidades.append(cantidad)

plt.figure(figsize=(12, 6))
plt.bar(clases_tomate, cantidades)

plt.xticks(rotation=90)
plt.ylabel("Número de imágenes")
plt.title("Distribución de imágenes por clase")

plt.show()
```

```python
import random

plt.figure(figsize=(15, 20))

indice = 1

for clase in clases_tomate:
    ruta_clase = os.path.join(dataset_path, clase)

    imagenes = [
        archivo for archivo in os.listdir(ruta_clase)
        if archivo.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    imagen_nombre = random.choice(imagenes)

    imagen = Image.open(
        os.path.join(ruta_clase, imagen_nombre)
    )

    plt.subplot(5, 2, indice)
    plt.imshow(imagen)
    plt.title(clase)
    plt.axis("off")

    indice += 1

plt.tight_layout()
plt.show()
```
